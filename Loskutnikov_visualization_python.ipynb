### Задание 1 (0.5 балла)

Используя `numpy` и `matplotlib`, начнём с тренировки: постройте график функций $sin(x)$ и $cos(x)$ на интервале от $-\pi$ до $\pi$, используя 100 точек. Добавьте легенду и сетку.. Используйте для линий разные цвета, одна из них пусть будет пунктирной. Результат в этом и следующих заданиях на построение графиков должен получиться примерно как на картинке. Содержание подписей не должно обязательно совпадать, однако осмысленные подписи должны присутствовать на графиках.

````python
x = np.arange(-np.pi, np.pi, 0.01)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize=(10,7),dpi=70)
plt.grid()

plt.plot(x, y_sin, linewidth=1, color='c')
plt.plot(x, y_cos, linestyle='--', linewidth=1, color='m')

plt.title("График синуса и косинуса")
plt.xlabel('x')
plt.ylabel('y')

plt.legend(['y=sin(x)', 'y=cos(x)'], loc='upper left')

plt.show();
````

````python
data_by_risk = data.groupby(['status','credit_risk']).size().unstack().reset_index()
data_by_risk
````

````python
fig, (ax1, ax2) = plt.subplots(1,2,figsize=(15,5),dpi=70)

# создадим первый график
data_by_risk.plot(
    kind='barh',
    x='status',
    y='good',
    width=0.8,
    ax=ax1,
    color='green',
    legend=False
    )

ax1.set_title('Положительный кейс',size=10)
ax1.set_xlabel('Количество кейсов')
ax1.set_ylabel('Статус текущего счёта')
ax1.set_xlim(0,370)
ax1.invert_xaxis()

# создадим второй график
data_by_risk.plot(
    kind='barh',
    x='status',
    y='bad',
    width=0.8,
    ax=ax2,
    color='red',
    legend=False
    )

ax2.set_title('Отрицательный кейс',size=10)
ax2.set_xlabel('Количество кейсов')
ax2.set_ylabel('')
ax2.set_xlim(0,370)
ax2.yaxis.tick_right()

# установим наименование фигуры
fig.suptitle('Статус текущего счёта', size=10)

plt.tight_layout()
plt.show();
````

Теперь постройте гистограмму для сравнения количества положительных и отрицательных кейсов в зависимости от размера сберегательного счёта с помощью seaborn. Какой способ вам больше нравится с точки зрения трудоёмкости написания кода? А с точки зрения восприятия результата?

````python
# создадим датафрейм для графика
data_by_savings = data.groupby(['savings', 'credit_risk']).size().\
    reset_index(name='count')
data_by_savings
````

````python
# создадим лист категорий для сортировки как на рисунке
list_of_savings = [
    'unknown/no savings account',
    '... <  100 DM',
    '100 <= ... <  500 DM',
    '... >= 1000 DM',
    '500 <= ... < 1000 DM',
    ]
# создадим оболочку фигуры для корректного отображения подписей по оси Х
fig = plt.figure(figsize=(12,6), dpi=70)

# создадим график
sns.barplot(
    data=data_by_savings,
    x='savings',
    y='count',
    hue='credit_risk',
    order=list_of_savings,
    hue_order=['good','bad'],
    palette={'good': 'green', 'bad': 'red'},
    legend='auto'
)

plt.title('Количество положительных и отрицательных кейсов\nв зависимости от размера сберегательного счета', fontsize=14)
plt.xlabel('Размер сберегательного счёта')
plt.ylabel('Количество кейсов')
plt.ylim(0,410);
````

###Вывод по Заданию 3:

На графике можно заметить, что положительных кейсов больше, чем отрицательных.

3. Заёмщики с низким сберегательным балансом могут быть рискованными для кредитора.

````python
df_netflix = pd.read_csv(NETFLIX_TITLES)
df_netflix.head()
````

````python
print('Количество пропусков в колонке director составляет:',
      df_netflix['director'].isna().sum())
````

````python
top_5_nan = df_netflix.query('director.isna() & type == "Movie"')\
    .groupby('country').size()\
    .reset_index(name='count')\
    .sort_values('count', ascending=False).head(5).reset_index(drop=True)
top_5_nan.style.bar(subset=['count'], color='#67A5EB')
````

````python
fig = plt.figure(figsize=(10,6))

sns.barplot(
    data=top_5_nan,
    x='country',
    y='count',
    hue='country'
)

plt.title('Топ 5 стран по количеству фильмов\nс пропусками в столбце director')
plt.xlabel('Страна')
plt.ylabel('Количество фильмов')

plt.tight_layout()
plt.show();
````

### Задание 6 (1 балл)

Будем работать со столбцом duration. Одни данные даны в минутах (для фильмов (Movie)), а другие в сезонах (для сериалов (TV shows)). Создайте два новых столбца min (для тех, у кого продолжительность в минутах) и season (для тех, у кого продолжительность в сезонах), в которых будут извлечены данные из столбца duration.

Постройте subplots, где на первом графике гистрограмма распределения продолжительности фильмов, а на втором – гистрограмма распределения продолжительности сериалов.

Добавьте на графики:

+ Заголовки жирным шрифтом
+ Подписи осей

````python
# создаём столбцы min и season
df_netflix['min'] = df_netflix[df_netflix['type'] == "Movie"]['duration']
df_netflix['season'] = df_netflix[df_netflix['type'] == "TV Show"]['duration']

movies_duration = df_netflix.groupby('min').size().reset_index(name='count')
movies_duration['min_num'] = movies_duration['min'].str.extract('(\d+)').astype(int)
movies_duration = movies_duration.sort_values('min_num')

tvshows_duration = df_netflix.groupby('season').size().reset_index(name='count')
tvshows_duration['season_num'] = tvshows_duration['season']\
    .str.extract('(\d+)').astype(int)
tvshows_duration = tvshows_duration.sort_values('season_num')
````

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# построим график продолжительности фильмов
sns.histplot(
    data=movies_duration,
    x='min_num',
    ax=ax1,
    bins='auto'
)

ax1.set_title('Распределение продолжительности фильмов', fontweight='bold')
ax1.set_xlabel('Продолжительность (минуты)')
ax1.set_ylabel('Количество фильмов')

# построим график распределения продолжительности сериалов
sns.barplot(
    data=tvshows_duration,
    x='season',
    y='count',
    ax=ax2
)

ax2.set_title('Распределение продолжительности сериалов', fontweight='bold')
ax2.set_xlabel('Продолжительность (сезоны)')
ax2.set_ylabel('Количество сериалов')
ax2.tick_params(axis='x', rotation=45)


plt.tight_layout()
plt.show();
````

На графике распределения продолжительности фильмов можно заметить, что большая часть фильмов находится на промежутке от 3 до 200 минут. Фильмов с большей продолжительность в данных почти нет.

На графике распределения продолжительности сериалом видно, что больше всего значений сериалов с продолжительность в 1 сезон (1600 сериалов). Сериалов с 2 сезонами в 4 раза меньше. С увеличением количество сезонов, количество сериалов снижается.
